## Content Prioritization Using SEO Metrics and Decision Tree Classification

### Abstract

This project investigates whether SEO performance metrics can be used to prioritize content optimization opportunities. A Decision Tree Classifier was applied to search volume, CTR, average position, engagement rate, and impression data from an anonymized FlyRank internship dataset. Grouped validation using `GroupKFold` was used to reduce leakage risk between clients. The model was compared with a rule-based baseline and produced identical accuracy on the grouped split, indicating that the current target definition is strongly aligned with the baseline logic. The project demonstrates a reproducible and interpretable workflow for SEO content prioritization while emphasizing validation design, leakage awareness, and careful interpretation of results.


## 1. Question

*The research question and the decision it supports.*

## Research Question

Can SEO performance metrics such as **search volume, click-through rate (CTR), average position, engagement rate, and impressions** be used to identify webpages that should be prioritized for content optimization?

### Decision Supported

The project supports a **content-prioritization decision**. The goal is to help a content or SEO team decide **which pages should be reviewed and updated first**, not to automate publishing or make final business decisions without human review.

## 3. Methodology

*Assumptions, features, label definition, baseline, validation design, leakage checks.*

## Methodology

### Model

The selected model was **Decision Tree Classifier (`max_depth=4`)** because it is simple, interpretable, and suitable for tabular SEO data.

### Baseline

A rule-based baseline from Week 4 was retained for comparison:

* high search volume,
* low CTR,
* average position greater than 10.

### Validation Design

To reduce the risk of leakage between clients, the dataset was split using **`GroupKFold(n_splits=5)`** with `client_id` as the grouping variable.

### Split Used

* **Training rows:** 22,992
* **Testing rows:** 7,008

### Leakage Check

The validation audit from Week 6 confirmed that grouping by client is a safer design than a purely random split because pages from the same client may share similar optimization patterns and performance characteristics.

### Interpretation

Both models produced identical accuracy on the grouped validation split. This suggests that the current target variable is strongly determined by the same threshold conditions used in the baseline rule.

The result should therefore be interpreted carefully: it shows that the Decision Tree can reproduce the baseline behavior, but it does **not** demonstrate that the model has learned substantially more useful patterns than the rule-based approach.


## 5. Limitations

*What this work cannot claim.*

## Limitations

This work has several important limitations:

* The target variable was created from manually defined thresholds rather than real editorial or business outcomes.
* Accuracy alone is not sufficient to measure SEO impact.
* The dataset represents anonymized internship data and may not reflect all production SEO scenarios.
* The identical baseline and model accuracy indicates that the label definition is closely aligned with the baseline logic.
* The system should be treated as a **directional decision-support workflow**, not as a production-ready autonomous SEO optimization system.
