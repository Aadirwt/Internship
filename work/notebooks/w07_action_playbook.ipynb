## 1. Ranked actions + reason codes

## Ranked actions and reason codes

| Priority | Content archetype                           | Recommended action                    | Reason code |
| -------- | ------------------------------------------- | ------------------------------------- | ----------- |
| 1        | High search volume + low CTR + poor ranking | Refresh title and meta description    | RC-01       |
| 2        | Good engagement + average position > 10     | Add internal links from related pages | RC-02       |
| 3        | High impressions + low clicks               | Improve search snippet and headings   | RC-03       |
| 4        | Declining trend pages                       | Update outdated sections and examples | RC-04       |
| 5        | Stable high-performing pages                | Monitor only; no immediate changes    | RC-05       |

### Reason code meanings

* **RC-01:** Visibility opportunity exists because ranking is outside the top search positions.
* **RC-02:** Engagement suggests the content satisfies user intent, so authority and discoverability should be improved.
* **RC-03:** The page is being shown in search results but is not attracting enough clicks.
* **RC-04:** Traffic decay suggests the content may be outdated or less competitive than newer pages.
* **RC-05:** Current performance is acceptable, so resources should be focused elsewhere.

## 3. Human review + the no-go list

*What a person must check before acting. What should never be automated.*

## Human review and no-go list

### Human review checklist

Before applying any recommendation:

* Confirm the target keyword is still relevant.
* Check that traffic changes are not caused by seasonality.
* Review competitor pages for recent updates.
* Ensure analytics data is complete and recent.

### No-go automation list

The following actions should **not** be automated:

* Publishing rewritten content directly to the website.
* Changing page titles for branded or legal pages.
* Removing content based only on low traffic.
* Creating backlinks automatically.
* Updating medical, financial, or legal content without expert review.

These actions require human judgment because incorrect changes could reduce search performance or introduce factual errors.

## 4. Monitoring / retrain triggers

*What would tell you the recommendations went stale?*

## Monitoring and retrain triggers

The model should be reviewed periodically to ensure that recommendations remain useful.

### Monitoring triggers

* Average CTR changes by **more than 15%** for two consecutive weeks.
* Average search position changes by **more than 3 positions**.
* Engagement rate drops below **50%** for a significant group of pages.
* A large number of new pages are added to the dataset.

### Retrain triggers

Retraining should be considered when:

* The feature distribution changes substantially (data drift).
* New SEO metrics become available.
* The model accuracy drops noticeably compared with the Week 5 baseline.
* The business begins targeting a different content strategy or audience segment.

The retraining process should reuse the same validation approach introduced in Week 6 to avoid leakage and maintain comparable results.
